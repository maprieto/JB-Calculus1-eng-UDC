# The indefinite integral

## Definition of antiderivative

The antiderivative of a function is the expression obtained through the reverse process of differentiation. For this reason, an antiderivative is also known as an anti-derivative. In fact, a function does not have a single antiderivative, but rather a family of functions. Let us see this with an example.

If we have the function $f(x)=x^2$, we know that its derivative is obtained in two steps: i) multiply the function by the exponent and ii) decrease the exponent by one unit, that is, $f'(x)=2x$. Therefore, to compute an antiderivative of $f$ we proceed in the opposite direction: i) increase the exponent by one unit and ii) divide by the exponent (the new one obtained in step i)). In this way we get $F(x) = \dfrac{x^3}{3}$, and we can check that applying the derivative to $F$ gives back our initial function $f$. 

As we have mentioned, the antiderivative is not unique. Thus we can add any constant to $F(x)$ and its derivative will still be the original function $f$ (the derivative of a constant is 0). Therefore, in this case, the family of antiderivatives is given by $F(x) = \dfrac{x^3}{3} +C$, where $C$ is any constant.

Let us now state the formal definitions of antiderivative.


````{prf:definition} Antiderivative
:label: def_primitiva 
:nonumber: 
Let $f:I\rightarrow \mathbb{R}$. We say that $F$ is an antiderivative of $f$ on $I$ if

$$ F'(x) = f(x), \quad \forall x \in I.$$
````

````{prf:theorem} 
:label: th_primitivas
If $F$ and $G$ are two antiderivatives of the same function $f$ on an interval $I$, then

$$ \exists k \in \mathbb{R} \mbox{ such that } F(x)=G(x)+k, \quad \forall x \in I.$$
````

As a consequence of the previous theorem, once one antiderivative $F$ of $f$ is known, all the others are obtained just by adding a constant to the known antiderivative.

## Definition of the indefinite integral
Indefinite integration is the process through which the antiderivatives of a function are obtained. The most common notation for the indefinite integral of the function $f$ is $\int f(x) \mathrm{d}x$, which is the notation we use in its definition.

````{prf:definition} Indefinite integral
:label: def_primitiva 
:nonumber: 

Given a function $f:I\rightarrow \mathbb{R}$, the indefinite integral of $f$ is the set of all antiderivatives of $f$, and it is written as

$$\int f(x)\, \mathrm{d}x = \{ F : F'(x)=f(x), \quad \forall x \in I\}.$$
````

Using {prf:ref}`th_primitivas`, it follows immediately that if we know one antiderivative $F$ of $f$, then its indefinite integral is

$$\int f(x) \, \mathrm{d}x = \{ F(x) +k, \quad \forall k \in \mathbb{R}\}.$$

Among the properties of the integral is linearity, that is:

````{prf:property} Linearity of indefinite integration
:label: prop_linealidad_int_indef 
:nonumber: 
Let $f:I\rightarrow \mathbb{R}$ and $g:I\rightarrow \mathbb{R}$ be functions. Then

* $\displaystyle\int \left( f(x)+ g(x)\right) \, \mathrm{d}x = \int f(x) \, \mathrm{d}x + \int g(x) \, \mathrm{d}x$,
* $\displaystyle\int \alpha f(x) \, \mathrm{d}x  = \alpha \int f(x) \, \mathrm{d}x, \quad \forall \alpha \in \mathbb{R}$.
````

Let us now look at some immediate integrals.

````{prf:property} Immediate integrals
:label: prop_integrales_inmediatas 
:nonumber: 

* $\displaystyle\int x^m \, \mathrm{d}x = \dfrac{1}{m+1}x^{m+1} + C, \quad m \neq -1$,
* $\displaystyle \int \dfrac{1}{x} \, \mathrm{d}x = \ln|x| + C$,
* $\displaystyle \int e^x \, \mathrm{d}x = e^x + C; \quad \int a^x \, \mathrm{d}x = \dfrac{a^x}{\ln a} + C, \quad a>0, a\neq1$,
* $\displaystyle \int \sin(x) \, \mathrm{d}x = -\cos(x) + C; \quad \int \cos(x) \, \mathrm{d}x = \sin(x) + C$,
* $\displaystyle \int \dfrac{1}{\cos^2(x)} \, \mathrm{d}x = \tan(x) + C$,
* $\displaystyle \int \dfrac{1}{1+x^2} \, \mathrm{d}x = \arctan(x) + C$,
* $\displaystyle \int \dfrac{1}{\sqrt{1-x^2}} \, \mathrm{d}x = \arcsin(x) + C$.
````

Would you like to practice a bit with these so-called *immediate integrals*? You can do so here: 
* https://www.matesfacil.com/ejercicios-resueltos-integrales-inmediatas.htm
* https://thales.cica.es/rd/Recursos/rd97/Problemas/54-1-p-INM.HTML


## Main methods for obtaining indefinite integrals
The most common methods for obtaining the indefinite integral of a function are integration by parts, integration by substitution, and integration by decomposition into rational fractions.

### Integration by parts
Integration by parts is used to find the integral of some functions $f$ that are the product of two others in a simpler way. Let us state this method.

````{prf:property} Integration by parts for the indefinite integral
:label: prop_integracion_ind_partes 
:nonumber: 
If $u'$ and $v'$ are continuous functions, then

$$ \int u(x)v'(x) \, \mathrm{d}x = (uv)(x) - \int v(x)u'(x)\, \mathrm{d}x,  $$
or equivalently

$$ \int u \, dv = uv - \int v \, du.  $$
````

It is convenient to use this method for products of:
* polynomial (acting as $u$) times exponential (acting as $dv$), 
* polynomial ($u$) times sine or cosine ($dv$),
* inverse trigonometric or logarithmic function ($u$) times polynomial ($dv$),
* exponential ($u$ or $dv$) times sine or cosine ($dv$ or $u$).

Let us see an example of how to apply integration by parts.

````{prf:example} 
:label: ex_indefinida_partes 
:nonumber: 

Compute the indefinite integral $\int xe^{x-1} \,\mathrm{d}x$.

We must keep in mind that we want to simplify the integrand so that the new integral to be computed is direct, or at least simpler than the original one. Therefore it is convenient to take $u=x$ and $dv=e^{x-1}dx$, so after differentiating and integrating we get $du = dx$ and $v=e^{x-1}$, respectively. In this way,

$$ \int xe^{x-1} \,\mathrm{d}x = xe^{x-1} - \int e^{x-1} dx,  $$

where the remaining integral is already immediate:

$$ \int xe^{x-1} \,\mathrm{d}x = xe^{x-1} -  e^{x-1} + C = (x-1)e^{x-1} + C.  $$
````

Let us look at another example where we can use this method.

````{prf:example} 
:label: ex_indefinida_partes_2
:nonumber: 

Find the following indefinite integrals:

* a) $\int e^x \sin(x) \,\mathrm{d}x$ 

    In this case we take $u=\sin(x)$ and $dv=e^x dx$. Therefore $ du = \cos(x)$ and $v = e^x$, and applying integration by parts we obtain

    $$ \int e^x \sin(x) \,\mathrm{d}x = e^x \sin(x) - \int e^{x} \cos(x) dx.  $$

    Now we must use integration by parts again to solve the new integral appearing on the right-hand side. We take $u=\cos(x)$ and $dv=e^x dx$, so $ du = -\sin(x)$ and $v = e^x$, and hence

    \begin{eqnarray*}
    \int e^x \sin(x) \,\mathrm{d}x &=& e^x \sin(x) - \int e^{x} \cos(x) dx \\
    &=& e^x\sin(x) -e^x\cos(x) - \int e^{x} \sin(x) dx. 
    \end{eqnarray*}

    After applying integration by parts a second time, we have obtained the same integral we started with, so we can find its value by solving for it in the previous equality:

    $$ 2\int e^x \sin(x) \,\mathrm{d}x = e^x\sin(x) -e^x\cos(x).$$

    Therefore we obtain


    $$ \int e^x \sin(x) \,\mathrm{d}x = \dfrac{ e^x}{2}\left(\sin(x) -\cos(x)\right) +C.$$

* b) $\int \ln(x) \,\mathrm{d}x$ 

    We take $u=\ln(x)$ and $dv=dx$. Then $du=\dfrac{1}{x}$ and $v=x$, so we obtain

    $\int \ln(x) \,\mathrm{d}x = x\ln(x) - \int 1dx = x\ln(x) -x +C = x(\ln(x) -1) +C. $
````

### Integration by substitution
This integration method can be thought of as the reverse of the chain rule in differentiation. The goal of this method is to turn the function to be integrated into something simpler.

````{prf:property} Integration by substitution
:label: prop_integ_cambio_variable 
:nonumber: 

Let $f:[a,b] \rightarrow \mathbb{R}$ be integrable and $\varphi:[\alpha,\beta] \rightarrow \mathbb{R}$ injective, with continuous derivative and such that $\varphi([\alpha,\beta])\subset [a,b]$. Then

$$\int f(x) \,\mathrm{d}x =  \int f(\varphi(t))\varphi'(t) \,\mathrm{d}t. $$
````

Let us look at an example where substitution is used.

````{prf:example} 
:label: ex_indefinida_cambio_variable 
:nonumber: 

* a) To compute $\displaystyle\int \dfrac{(\ln(x))^3}{x}  \,\mathrm{d}x $ we may take $t = \ln(x)$, and differentiating the previous expression gives $dt=\dfrac{dx}{x}$. Therefore, after making the substitution in the original integral, we obtain

    $$\int \dfrac{(\ln(x))^3}{x}  \,\mathrm{d}x = \int t^3 \,\mathrm{d}t,$$

    which is already an immediate integral:

    $$\int t^3 \,\mathrm{d}t = \dfrac{t^4}{4} + C.$$

    Finally, undoing the substitution, we obtain

    $$\int \dfrac{(\ln(x))^3}{x}  \,\mathrm{d}x = \dfrac{(\ln(x))^4}{4} + C.$$


* b) To compute $\displaystyle\int \dfrac{1}{1+\sqrt{x}}  \,\mathrm{d}x $ we may take $t = \sqrt{x}$, which gives 
  $dt=\dfrac{dx}{2\sqrt{x}} = \dfrac{dx}{2t}$. In other words, $dx = 2t dt$, so after making the substitution we obtain

    $$\int \dfrac{1}{1+\sqrt{x}}  \,\mathrm{d}x = \int \dfrac{2t}{1+t} \,\mathrm{d}t,$$

    which is still not an immediate integral. So we can make another substitution: let $z= 1+t$, then $dz=dt$, and

    $$ \int \dfrac{2t}{1+t} \,\mathrm{d}t = \int \dfrac{2z-2}{z} \,\mathrm{d}z = \int 2 \,\mathrm{d}z-\int \dfrac{2}{z} \,\mathrm{d}z = 2z -2\log|z| +C.$$

    Now we must undo the two substitutions made above, starting with the last one:

    $$ \int \dfrac{2t}{1+t} \,\mathrm{d}t = 2t+2 -2 \ln|t+1| +C,$$

    and then the first substitution:

    \begin{eqnarray*}
    \int \dfrac{1}{1+\sqrt{x}}  \,\mathrm{d}x &=& 2\sqrt{x} +2 -2\ln|\sqrt{x}+1| +C \\
    &=& 2\left(1+\sqrt{x} -\ln(\sqrt{x}+1)\right) +C,
    \end{eqnarray*}

    where the absolute value has been removed because $\sqrt{x}+1>0$.
````

### Integration by decomposition into rational fractions
This method is used to integrate rational functions $f(x)=\dfrac{p_m(x)}{q_n(x)}$ in which the degree of the numerator ($m$) is smaller than the degree of the denominator ($n$). This is always possible because, if this is not the case, polynomial division can be carried out. That is, if $m\ge n$, then $\dfrac{p_m(x)}{q_n(x)} = c_r(x) + \dfrac{r_s(x)}{q_n(x)}$, where the quotient is a polynomial that can be integrated directly, and the new function to be integrated by this method corresponds to the remainder of the division divided by the divisor, $\dfrac{r_s(x)}{q_n(x)}$, with $s<n$. 

````{prf:property} 
:label: prop_integ_racionales 
:nonumber: 

Suppose we have a rational function $f(x)=\dfrac{p_m(x)}{q_n(x)}$ with $m<n$, and $\displaystyle q_n(x)=a_{n_a}(x)\prod_{i=1}^{R} (x-\alpha_i)^{s_i}$, where $\alpha_i$ are the distinct real roots of multiplicity $s_i$ of $q_n(x)$, with $i=1,2,\dots,R$, and $a_{n_a}(x)$ is a polynomial with no real roots, with $n_a<n$, such that $n_a +s_1+s_2 +\dots+s_R = n$. Then we have

\begin{eqnarray*}
\dfrac{p_m(x)}{q_n(x)} &=& \dfrac{b_{n_b}(x)}{a_{n_a}(x)} + \dfrac{k_1}{x-\alpha_1} + \dots+ \dfrac{k_{s_1}}{(x-\alpha_1)^{s_1}} + \\
&& + \dots + \dfrac{k_{n_{r-1}+1}}{x-\alpha_r} + \dots+ \dfrac{k_{n_{r-1}+s_r}}{(x-\alpha_r)^{s_r}},
\end{eqnarray*}

where $n_b < n_a$ and $n_{r-1}=s_1+s_2+\dots+s_{r-1}$. In this case, the integral of $f$ can be decomposed as

\begin{eqnarray*}
\int f(x) \,\mathrm{d}x &=&  \int \dfrac{b_{n_b}(x)}{a_{n_a}(x)}\,\mathrm{d}x  + \int \dfrac{k_1}{x-\alpha_1}\,\mathrm{d}x  + \dots+ \int \dfrac{k_{s_1}}{(x-\alpha_1)^{s_1}} \,\mathrm{d}x + \\
&& + \dots + \int \dfrac{k_{n_{r-1}+1}}{x-\alpha_r}\,\mathrm{d}x  + \dots+ \int \dfrac{k_{n_{r-1}+s_r}}{(x-\alpha_r)^{s_r}} \,\mathrm{d}x .
\end{eqnarray*}
````

Let us look at some practical examples.

````{prf:example} 
:label: ex_integ_racionales 
:nonumber: 

* **Example 1:** Compute $\int \dfrac{2x}{x^3 -x^2 +x -1} \,\mathrm{d}x$.

    We can factor the denominator as $(x^2+1)(x-1)$ and therefore

    $$\dfrac{2x}{x^3 -x^2 +x -1}= \dfrac{ax+b}{x^2+1}+\dfrac{k_1}{x -1},$$

    where we must determine the values of the parameters $a, \ b$ and $k_1$. To do this, we add the fractions:

    $$\dfrac{ax+b}{x^2+1}+\dfrac{k_1}{x -1} = \dfrac{(a+k_1)x^2 + (b-a)x +(k_1-b)}{(x^2+1)(x-1)}.$$

    Since we want the denominator to match the original one, we must have $(a+k_1)x^2 + (b-a)x +(k_1-b)=2x$. Therefore $a+k_1 = 0$, $b-a = 2$, and $k_1-b=0$, from which we obtain $b=k_1=1$ and $a=-1$. Hence

    \begin{eqnarray*}
    \int \dfrac{2x}{x^3 -x^2 +x -1} \,\mathrm{d}x &=& \int \dfrac{-x+1}{x^2+1}\,\mathrm{d}x+\int \dfrac{1}{x -1}\,\mathrm{d}x \\
    &=&  \int \dfrac{1}{x^2+1}\,\mathrm{d}x - \int \dfrac{x}{x^2+1}\,\mathrm{d}x+\int \dfrac{1}{x -1}\,\mathrm{d}x. 
    \end{eqnarray*}

    The last three integrals are already immediate (in the second one, a substitution could be made if it is not immediately obvious). In this way we obtain

    $$\int \dfrac{2x}{x^3 -x^2 +x -1} \,\mathrm{d}x =  \arctan(x) -\dfrac{1}{2}\ln(x^2+1) + \ln|x-1| + C. $$

* **Example 2:** Compute $\int \dfrac{x^2+3x+2}{x^3-3x^2+4} \,\mathrm{d}x$.

    In this case $x^3-3x^2+4=(x+1)(x-2)^2$, therefore 

    $$\dfrac{x^2+3x+2}{x^3-3x^2+4}= \dfrac{k_1}{x +1} + \dfrac{k_2}{x -2} + \dfrac{k_3}{(x -2)^2}.$$

    Adding the fractions gives

    \begin{eqnarray*}
    \dfrac{k_1}{x +1} &+& \dfrac{k_2}{x -2} + \dfrac{k_3}{(x -2)^2} = \\
    &&\dfrac{(k_1+k_2)x^2 + (k_3 -k_2 -4k_1)x + (k_3-2k_2+4k_1)}{(x+1)(x-2)^2}.
    \end{eqnarray*}

    For the numerators to match, we must have $k_1+k_2=1$, $k_3 -k_2 -4k_1=3$, and $k_3-2k_2+4k_1=2$, from which we obtain $k_1=0$, $k_2=1$, and $k_3=4$. Thus

    $$\int \dfrac{x^2+3x+2}{x^3-3x^2+4} \,\mathrm{d}x = \int \dfrac{0}{x +1} \,\mathrm{d}x + \int\dfrac{1}{x -2} \,\mathrm{d}x+ \int\dfrac{4}{(x -2)^2} \,\mathrm{d}x,$$

    where the integrals on the right are already immediate. Note that $\dfrac{4}{(x -2)^2}=4(x-2)^{-2}$, and therefore

    $$\int \dfrac{x^2+3x+2}{x^3-3x^2+4} \,\mathrm{d}x  = \ln|x-2| -\dfrac{4}{x-2}+ C.$$
````
