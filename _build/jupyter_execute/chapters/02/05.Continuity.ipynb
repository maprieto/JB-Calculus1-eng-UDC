# Continuity

## Continuity at a point

````{prf:definition}
:label: def_continuidad
:nonumber:

Let $f:A\subset\mathbb{R}\rightarrow\mathbb{R}$, $x_0\in A$. We will say that **$f$ is continuous at $x_0$**
if and only if $\displaystyle\lim_{x\to x_0} f(x)$ exists and that limit is $f(x_0)$.

That is, $f$ is continuous at $x_0$ if and only if

$$
\forall\epsilon>0, \exists\delta>0 \Big/
\left|x-x_0\right|<\delta\Rightarrow\left|f(x)-f(x_0)\right|<\epsilon.
$$
````

Let us observe that, with respect to the definition of limit, there is a small difference. Before we wrote

$$
0< \left|x-x_0\right|<\delta,
$$
whereas in the case of continuity we simply write

$$
\left|x-x_0\right|<\delta.
$$

Why? When we talk about limit, we are interested in the point $x_0$ not taking part in
the definition, so that we can have limits of functions that are not even defined
at $x_0$. Hence the condition $0<|x-x_0|$ (obviously, $0=|x-x_0|$ if and only if $x=x_0$).
However, in the case of continuity, including the point $x_0$ is natural. Moreover, for
continuity it cannot spoil anything, because if $x=x_0$ then $|f(x)-f(x_0)|$ is
$0$, so it will always be smaller than any positive $\epsilon$.

In the following figure we show graphically the possible cases in which continuity fails.
The first two ($\not\exists f(x_0)$, $\displaystyle\not\exists\lim_{x\to x_0}f(x)$) are called **essential discontinuities**, whereas the third case
($\displaystyle \lim_{x\to x_0}f(x)\not=f(x_0)$) is a **removable discontinuity**.

<img src="../../images/cap3_discontinuities.png" width="500"/>

A touch of humor, taken from the Twitter account `@MathMatize`:

<img src="../../images/cap3_continuity_meme.jpg" width="500"/>

Some important properties of continuity are the following.

````{prf:property} Algebra of continuous functions
:label: prop_aritmetica_cont
:nonumber:

Let $f,g:A\subset\mathbb{R}\rightarrow\mathbb{R}$ be functions continuous at a point $x_0\in A$. Then
* $\lambda f$ is continuous at $x_0$, $\forall\lambda\in\mathbb{R}$,
* $f\pm g$ is continuous at $x_0$,
* $fg$ is continuous at $x_0$,
* if $g(x_0)\not=0$, $\frac{f}{g}$ is continuous at $x_0$.
````

````{prf:property}
:label: prop_comp_cont
:nonumber:

**The composition of continuous functions is a continuous function**. That is,

$$
\left.\begin{array}{l}
f:A\subset\mathbb{R}\rightarrow\mathbb{R} \text{ continuous at } x_0\in A \\
g:f(A)\rightarrow\mathbb{R}\text{ continuous at }f(x_0)
\end{array}\right\}\Longrightarrow
g\circ f:A\subset\mathbb{R}\rightarrow\mathbb{R} \text{ continuous at }x_0\in A.
$$
````

````{prf:property}
:label: prop_lim_cont
:nonumber:

**The limit commutes with continuous functions**. That is, let $f$ and $g$ be functions such
that $\displaystyle\lim_{x\to x_0}f(x)=l\in\mathbb{R}$ exists and $g$ is a function continuous at $l$.
Then

$$
\lim_{x\to x_0}g\left(f(x)\right)=g\left(\lim_{x\to x_0}f(x)\right)=g(l).
$$
````

Thus, for example,

$$
\lim_{x\to 1}\sin\left(\frac{\ln x+\cos^2 x}{e^x+2x}\right)=\sin\left(\lim_{x\to 1}\frac{\ln
x+\cos^2 x}{e^x+2x}\right),
$$
since the sine function is continuous.

````{prf:definition}
:label: def_raiz
:nonumber:

Let $f:A\subset\mathbb{R}\rightarrow\mathbb{R}$. We will say that $x_0\in A$ is a **root** of $f$ if
$f(x_0)=0$.
````

## Bolzano's and Weierstrass's theorems

````{prf:theorem} Bolzano's theorem
:label: th_Bolzano
:nonumber:

Let $f:[a,b]\rightarrow\mathbb{R}$ be a continuous function on $[a,b]$ such that $f(a)f(b)<0$.
Then there exists $x_0\in(a,b)$ such that $f(x_0)=0$.

<img src="../../images/cap3_Bolzano1y2.png" width="500"/>
````

````{prf:theorem} Weierstrass's theorem
:label: th_Weierstrass
:nonumber:

Let $f:[a,b]\rightarrow\mathbb{R}$ be a continuous function on $[a,b]$. Then $f$ attains an absolute maximum and an absolute minimum on $[a,b]$. That is,

$$
\exists x_1, x_2\in[a,b]\Big/  f(x_1)\leq f(x)\leq f(x_2)\qquad\forall x\in[a,b].
$$
````
