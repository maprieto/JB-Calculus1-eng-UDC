# Fundamental Theorem of Calculus

## Statement of the theorem

In general, computing the value of a definite integral using geometry or limits of Riemann sums for infinitely refined partitions is a complicated task. In this section we present the Fundamental Theorem of Calculus, which gives us a way to compute many integrals. In fact, it reduces the integration problem to finding antiderivatives, that is, to the inverse problem of differentiation.

````{prf:theorem} Fundamental Theorem of Calculus
:label: th_fundamental 
:nonumber: 
Let $f \in {\cal R}[a,b]$. For $a \leq x \leq b$, let 

$$F(x) = \int_a ^x f(t) \,dt .$$
Then, $F \in {\cal C}[a,b]$. Moreover, if $f$ is continuous on $[a,b]$, then $F$ is differentiable on $[a,b]$
and $F^{\,\prime} (x) = f(x)$, $\forall x \in [a,b]$.

It can also be stated in the following way:

If $f : I \longrightarrow \mathbb R$ is continuous on $I$, then it has antiderivatives on $I$; 
one of them is the definite integral $F$ given by:

$$F(x) = \int_a ^x f(t) \,dt,$$
where $a \in I$ *is arbitrary*.
````

It would be good to practice computing $F$ a little in simple cases. Here are a couple of detailed examples:

* https://existelimite.blogspot.com/2012/11/un-ejercicio-sobre-el-teorema.html
* https://existelimite.blogspot.com/2012/12/un-ejercicio-en-flashback-sobre-el.html

## First consequence: differentiation of an integral function

Let us *play* a little with this theorem. We begin with a four-part exercise:

**EXERCISE:** 

1. Let $F(x)= \displaystyle \int_{1}^{x} t^2\sin(t)\, dt$. Compute $F'(x)$.

    According to the Fundamental Theorem, since the function inside the integral, $f(t)=t^2\sin(t)$, is continuous, the derivative of $F$ at $x$ coincides with the value of the integrand at that point:

    $$
     F'(x) = f(x) = x^2\sin(x).
    $$
2. Let $F(x)= \displaystyle \int_{1}^{x^3} t^2\sin(t)\, dt$. Compute $F'(x)$.

    In this case, we must combine the Fundamental Theorem with the chain rule to obtain:

    $$
     F'(x) = F'(x^3)3x^2 = f(x^3)3x^2 = \left(x^{3}\right)^{2}\sin(x^{3}) 3x^2 = 3x^{8}\sin(x^{3}).
    $$
3. Let $F(x)= \displaystyle \int_{\ln(x)}^{1} t^2\sin(t)\, dt$. Compute $F'(x)$.

    Now, first of all, we reverse the integral (changing its sign) so that the variable part appears in the upper limit,

    $$
    F(x) = \int_{\ln(x)}^{1} t^2\sin(t)\, dt = - \int_{1}^{\ln(x)} t^2\sin(t)\, dt = \int_{1}^{\ln(x)} - t^2\sin(t)\, dt,
    $$
    and now we proceed as in the previous part:

    $$
     F'(x) = F'(\ln(x))\frac{1}{x} = -(\ln(x))^2\sin(\ln(x))\frac{1}{x} = -\frac{(\ln(x))^2}{x}\sin(\ln(x)). 
    $$
4. Let $F(x)= \displaystyle \int_{\ln(x)}^{x^3} t^2\sin(t)\, dt$. Compute $F'(x)$.

    First we use additivity over intervals:

    $$
    F(x)= \displaystyle \int_{\ln(x)}^{x^3} t^2\sin(t)\, dt = \int_{\ln(x)}^{1} t^2\sin(t)\, dt + \int_{1}^{x^3} t^2\sin(t)\, dt,
    $$
    and then, combining the results of the previous parts:

    $$
    F'(x) = 3x^{8}\sin(x^{3}) - \frac{(\ln(x))^2}{x}\sin(\ln(x)).
    $$

These computations, carried out for a particular case, can be generalized easily:

````{prf:property} 
:label: prop_fundamental 
:nonumber: 
Let the function $F$ be given by the definite integral:

$$
F(x) = \int_{a(x)}^{b(x)} f(t) \, dt.
$$
Then, the derivative of $F$ with respect to $x$ is given by:

$$
F^{\,\prime}(x) = f(b(x)) \, b^{\prime}(x) - f(a(x)) \, a^{\prime}(x). 
$$
````

This opens up a world of possibilities. For example, if we know the derivative we can look for critical points, study concavity/convexity, or even compute the Taylor polynomial for functions defined from an integral!

## Second consequence: Barrow's rule

````{prf:property} Barrow's rule
:label: prop_Barrow 
:nonumber: 
If $f \in {\cal R}[a,b]$ and there exists an antiderivative $F$ of $f$ on $[a,b]$, then: 

$$
\int_a ^b f(x) \,dx = \left. \begin{array}{c} \, \\ \, \end{array} F(x) \right|_a^b = F(b) - F(a).
$$
````

````{prf:proof} 
Although it is not very common in this course, since it is very simple and we believe it is instructive, we are going to show the proof of this property... although only for the easiest case: we will assume that $f$ is a continuous function.

Suppose, then, that we can compute an antiderivative, $F$, of our function $f$ (using the techniques learned at school or in Section {ref}`sec:integral_indefinida`). 

On the other hand, thanks to the Fundamental Theorem of Calculus, we know another antiderivative of $f$:

$$
G(x) = \int_a ^x f(t) \,dt.
$$

Since we also know that two different antiderivatives of the same function differ by a constant, it follows that $F(x) = G(x) + C$. Then:

\begin{eqnarray*}
F(b) - F(a) &=& \left(G(b)+C\right) - \left(G(a)+C\right) = G(b) - G(a) = \int_{a}^{b} f(t)\, dt - \int_{a}^{a} f(t)\, dt \\
&=& \int_{a}^{b} f(t)\, dt - 0 = \int_{a}^{b} f(t)\, dt.
\end{eqnarray*}
````

Barrow's rule (we insist: a direct consequence of the Fundamental Theorem of Calculus) connects Riemann integration (something purely geometric) with the computation of antiderivatives (an algebraic operation): 
to compute the area under a positive function it is enough to find an antiderivative, evaluate it at the two endpoints, and subtract. Let us see an

````{prf:example} 
:label: ex_Barrow 
:nonumber: 
Compute the area under the curve $y=\cos x$ on $\left[-\frac{\pi}{2}, \frac{\pi}{2}\right]$.

**Solution:**

$$
A = \int_{-\frac{\pi}{2}}^{\frac{\pi}{2}} \cos x \, dx = \sin x \lvert_{-\frac{\pi}{2}}^{\frac{\pi}{2}} = \sin\left(\frac{\pi}{2}\right) - \sin\left(-\frac{\pi}{2}\right) = 1 - (-1) = 2.
$$
Thus, the area of the region is two square units.
````

## Integration by parts for the Riemann integral

````{prf:property} 
:label: prop_fundamental 
:nonumber: 

If $F$ and $G$ are two differentiable functions on $[a,b]$, and

$$\begin{cases} F^{\,\prime} = f \\ G^{\,\prime} = g \end{cases} \hspace{20mm} \text{on } [a,b],
$$ 
with $f$ and $g$ integrable on $[a,b]$, then

$$
\int_a ^b F(x) \,g(x) \,dx = F(b) \,G(b) - F(a) \,G(a) - \int_a ^b f(x) \,G(x) \,dx.
$$
````
