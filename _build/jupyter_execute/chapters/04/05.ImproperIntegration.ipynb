# Improper integration

Earlier we defined the Riemann integral of bounded functions on bounded intervals. Although it may seem paradoxical, in some cases it is also possible to compute the integral of a function that is not necessarily bounded on intervals that are not necessarily bounded; for example:
\begin{equation*}
\int_1^{\infty} \frac{dx}{x^3} \, , \qquad \qquad \int_{-1}^{1} \frac{dx}{x^2} \, .
\end{equation*}


These integrals are called **improper** and can be classified as:
- of the first kind, if the function is bounded but the interval is infinite,
- of the second kind, if the interval is finite but the function is unbounded on the interval,
- of the third kind, when neither the function nor the interval of integration is bounded.


## Improper integrals of the first kind

In these integrals, **the function is bounded but the interval of integration is infinite**.  

Let $f: [a,+\infty) \rightarrow \mathbb{R}$ be integrable on $[a, b]$ for every $b \geq a$. Then we define:
\begin{equation*}
\int_a^{+\infty} f(x) \, dx = \underset{b \rightarrow +\infty}{\lim} \int_a^b f(x) \, dx \, .
\end{equation*}
Its computation has two parts:
1. Compute the definite integral on a finite interval.
2. Let the upper endpoint of the interval tend to infinity.

Let us look at an example:
\begin{equation*}
I = \int_0^{+\infty} e^{-2x} \, dx \, .
\end{equation*}
The integrand is bounded and has an immediate antiderivative, but the interval of integration is not bounded to the right. Therefore, we compute our improper integral as
\begin{align*}
I = \int_0^{+\infty} e^{-2x} \, dx &= \underset{b \rightarrow +\infty}{\lim} \int_0^{b} e^{-2x} \, dx = \\
  &= \underset{b \rightarrow +\infty}{\lim} \left[ -\frac{1}{2} e^{-2x} \right]_0^b
   = \underset{b \rightarrow +\infty}{\lim} \left( -\frac{1}{2} \right) \left[ e^{-2b} - 1 \right]
   = -\frac{1}{2} \left[ 0 - 1 \right] = \frac{1}{2} \, .
\end{align*}

We have obtained a finite result. In these cases, we say that **the improper integral is convergent**. 


Similarly, we can solve an improper integral where the interval is unbounded on the left.

If $f: (-\infty, b] \rightarrow \mathbb{R}$ is integrable on $[a, b]$ for every $a \leq b$, we define:
\begin{equation*}
\int_{-\infty}^b f(x) \, dx = \underset{a \rightarrow -\infty}{\lim} \int_a^b f(x) \, dx \, .
\end{equation*}

For example,
\begin{equation*}
\int_{-\infty}^{-3} \frac{dx}{x^2} \, dx = \underset{a \rightarrow -\infty}{\lim} \int_a^{-3} \frac{dx}{x^2} \, dx
   = \underset{a \rightarrow -\infty}{\lim} \left[ -\frac{1}{x} \right]_a^{-3}
   =  \underset{a \rightarrow -\infty}{\lim} \left[ \frac{1}{3} + \frac{1}{a} \right]
   = \frac{1}{3} \, ,
\end{equation*}
or else
\begin{equation*}
\int_{-\infty}^1 e^{-x} \, dx = \underset{a \rightarrow -\infty}{\lim} \int_a^1 e^{-x} \, dx
   = \underset{a \rightarrow -\infty}{\lim} \left[ -e^{-x} \right]_a^1
   = \underset{a \rightarrow -\infty}{\lim} \left[ -e^{-1} + e^{-a} \right] = +\infty \, .
\end{equation*}
In the first case we again have a convergent integral (because the result is finite), whereas the second integral is **divergent**.


If the interval is unbounded at both ends, we must decompose the integral into two improper integrals where—in each of them—only one endpoint is unbounded:
\begin{align*}
\int_{-\infty}^{+\infty} \frac{dx}{x^2+1} 
    &= \int_{-\infty}^c \frac{dx}{x^2+1} + \int_c^{+\infty} \frac{dx}{x^2+1} = \\
    &= \underset{a \rightarrow -\infty}{\lim} \int_a^c \frac{dx}{x^2+1} 
     + \underset{b \rightarrow +\infty}{\lim} \int_c^b \frac{dx}{x^2+1} \\
    &= \underset{a \rightarrow -\infty}{\lim} \left[ \arctan (x) \right]_a^c 
     + \underset{b \rightarrow +\infty}{\lim} \left[ \arctan (x) \right]_c^b \\ 
    &= \left[ \arctan (c) - \underset{a \rightarrow -\infty}{\lim} \arctan (a) \right]
     + \left[ \underset{b \rightarrow +\infty}{\lim} \arctan (b) - \arctan (c) \right] \\
    &= \arctan (c) - \left( -\frac{\pi}{2} \right) + \frac{\pi}{2} - \arctan (c) = \pi \, .
\end{align*}
If the integral is well defined, the result will be independent of the point $c$ introduced.

````{prf:remark} 
:label: rem_impropia 
:nonumber: 
In this case, **if one of the integrals in the decomposition is divergent, then the whole integral is said to be divergent**, regardless of what happens to the other one. 
````


## Improper integrals of the second kind

We say that **an improper integral is of the second kind when the interval of integration is finite but the function to be integrated, $f$, is unbounded at some point of the interval**:
\begin{equation*}
\underset{x \rightarrow c}{\lim} f(x) = \pm \infty \qquad \text{ with } c \in [a,b] \, .
\end{equation*}

In these cases we must take limits in a neighborhood of the point (or points) where the function ceases to be bounded.

For example, consider the integral
\begin{equation*}
\int_2^4 \frac{dx}{\sqrt{x-2}} \, .
\end{equation*}
We can check that the integrand, $f(x) = \frac{1}{\sqrt{x-2}}$, tends to infinity when $x$ tends to $2$; since this point is precisely the left endpoint of the interval, it is enough to compute the integral on $[c, 4]$ and let $c$ tend to $2$:
\begin{align*}
\int_2^4 \frac{dx}{\sqrt{x-2}} &= \underset{c \rightarrow 2}{\lim} \int_c^4 (x-2)^{-1/2} \, dx = \\
    &= \underset{c \rightarrow 2}{\lim} \left[ \frac{(x-2)^{1/2}}{1/2} \right]_c^4
     = 2 \underset{c \rightarrow 2}{\lim} \left[ \sqrt{4-2} - \sqrt{c-2} \right] = 2 \sqrt{2} \, .
\end{align*}
It is done similarly if the function tends to infinity as we approach the right endpoint of the interval:
\begin{align*}
\int_2^4 \frac{dx}{(x-4)^{1/3}} &= \underset{c \rightarrow 4}{\lim} \int_2^c (x-4)^{-1/3} \, dx 
   = \underset{c \rightarrow 4}{\lim} \left[ \frac{(x-4)^{2/3}}{2/3} \right]_2^c \\
   &= \frac{3}{2} \underset{c \rightarrow 4}{\lim} \left[ (c-4)^{2/3} - (2-4)^{2/3} \right]
   = -\frac{3}{2} 2^{2/3} \, .
\end{align*}

And if the point where the function tends to infinity lies inside the interval of integration, it is enough to split the integral into a sum of two others such as those just described. For example:
\begin{align*}
\int_1^4 \frac{dx}{(x-2)^{1/3}} &= \int_1^2 \frac{dx}{(x-2)^{1/3}} + \int_2^4 \frac{dx}{(x-2)^{1/3}} \\
     &=\underset{b \rightarrow 2}{\lim} \int_1^b (x-2)^{-1/3} \, dx 
                                 + \underset{a \rightarrow 2}{\lim} \int_a^4 (x-2)^{-1/3} \, dx  \\
    &= \underset{b \rightarrow 2}{\lim} \left[ \frac{(x-2)^{2/3}}{2/3} \right]_1^b
     + \underset{a \rightarrow 2}{\lim} \left[ \frac{(x-2)^{2/3}}{2/3} \right]_a^4  \\
    &= \frac{3}{2} \underset{b \rightarrow 2}{\lim} \left[ (b-2)^{2/3} - (1-2)^{2/3} \right]
     + \frac{3}{2} \underset{a \rightarrow 2}{\lim} \left[ (4-2)^{2/3} - (a-2)^{2/3} \right]  \\
    &= \frac{3}{2} \left[ 0 - 1 \right] + \frac{3}{2} \left[ 2^{2/3} - 0 \right] 
     = \frac{3}{2} \left(\sqrt[3]{4} - 1 \right) \, .
\end{align*}


## Improper integrals of the third kind

This group includes **integrals in which both the interval and the function are unbounded**.  

To solve them, it is enough to combine the techniques we have seen in the previous sections.

Let us look at an example:
\begin{equation*}
\int_1^{+\infty} \frac{dx}{(x-4)^{2/5}} \, .
\end{equation*}
We see that the interval of integration is infinite and, moreover, the function to be integrated is unbounded at $x = 4$.  

To solve it, we decompose it into:
- an integral between $1$ and $4$ (improper of the second kind),
- an integral between $4$ and $6$ (improper of the second kind),
- an integral between $6$ and $+\infty$ (improper of the first kind).

We have chosen the point $6$, but we could have chosen any other point in $(4, +\infty)$. In this way,
\begin{align*}
\int_1^{+\infty} \frac{dx}{(x-4)^{2/5}} &= \int_1^4 \frac{dx}{(x-4)^{2/5}} 
       + \int_4^6 \frac{dx}{(x-4)^{2/5}} + \int_6^{+\infty} \frac{dx}{(x-4)^{2/5}} \\
    &= \underset{b \rightarrow 4}{\lim} \int_1^b (x-4)^{-2/5} \, dx 
     + \underset{a \rightarrow 4}{\lim} \int_a^6 (x-4)^{-2/5} \, dx \\
    & \hphantom{=} + \underset{c \rightarrow +\infty}{\lim} \int_6^c (x-4)^{-2/5} \, dx \\
    &= \underset{b \rightarrow 4}{\lim} \left[ \frac{(x-4)^{3/5}}{3/5} \right]_1^b
     + \underset{a \rightarrow 4}{\lim} \left[ \frac{(x-4)^{3/5}}{3/5} \right]_a^6
     + \underset{c \rightarrow +\infty}{\lim} \left[ \frac{(x-4)^{3/5}}{3/5} \right]_6^c \\
    &= \dfrac{5}{3} \left[ \underset{b \rightarrow 4}{\lim} \left( (b-4)^{3/5} - (-3)^{3/5} \right) 
                         + \underset{a \rightarrow 4}{\lim} \left( (6-4)^{3/5} - (a-4)^{3/5} \right) \right. \\
    & \hphantom{= \dfrac{5}{3}} \left. + \underset{c \rightarrow +\infty}{\lim} \left( (c-4)^{3/5} - 2^{3/5} \right) \right]  \\
    &= \dfrac{5}{3} \left[ 0 + 3^{3/5} + 2^{3/5} - 0 
                    + \underset{c \rightarrow +\infty}{\lim} (c-4)^{3/5} - 2^{3/5} \right] = +\infty \, .
\end{align*}
Therefore, in this example, we are dealing with a divergent integral.

Recall that it is enough for one of the pieces to be divergent for the whole integral to be considered divergent as well.


## Further information

You can find more information, and some examples/exercises, at
* https://es.wikipedia.org/wiki/Integral_impropia
* https://es.khanacademy.org/math/ap-calculus-bc/bc-integration-new/bc-6-13/a/improper-integrals-review

## Other exercises

In addition to the exercises you can find in the problem sets, we recommend taking a look at
* https://www.ehu.eus/~mtpalezp/mundo/ana1/EJERCICIOSINTEGRALIMPROPIA.pdf
* https://existelimite.blogspot.com/search/label/Integraci%C3%B3n%20impropia (they are a bit harder than usual, but... you can do it!).
