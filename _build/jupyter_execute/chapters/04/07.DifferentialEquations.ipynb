# Differential equations

## First definitions

````{prf:definition} 
:label: def_EDO 
:nonumber: 

* An **ordinary differential equation** (O.D.E.) is an equation in which an unknown function, $y=y(x)$, and its derivatives up to a certain order appear. 
* The **order** of an ordinary differential equation is the highest order of the derivatives that appear in the equation.
````
Let us look at some simple examples to get warmed up:

* First-order ODEs:
    * $y' = x^2 + 1$,
    * $\log(xy')=e^{xy}$,
    * $y^2 y' + x^2 y = 2$,
    * $T' + T = cos(5t)$,
* Second-order ODEs: 
    * $y''+5y'-6y = x^2 + 1$,
    * $mx'' - sx' = g$. 

````{prf:definition} 
:label: def_variables_dep_indep 
:nonumber: 

Inside an ODE we distinguish two types of variables:

* **Independent variable**. This is the variable that, within the relation under consideration, does not depend on any other. It is usually denoted by the letters $x$ or $t$ (although this is only a convention).
* **Dependent variable**. This is the variable whose values depend on those taken by the independent variables. It is usually denoted by the letter $y$ or by the initial, often capitalized, of the quantity it represents ($a$ for acceleration, $T$ for temperature, ...). 
````

Notice that, in an ODE with, for example, $x$ as the independent variable and $y$ as the dependent variable, when we write:

* $y$, we mean $y(x)$,
* $y'$, we mean $\displaystyle \frac{dy}{dx}$, *etc.*.


Once these first definitions are in place, from now on in these notes we will focus on first-order ODEs. We now give the following definition:

````{prf:definition} First-order ordinary differential equation
:label: def_EDO_orden_uno
:nonumber: 
A first-order ordinary differential equation is a differential equation of the form 

$$
y'\,=\,f(x,y)\,,
$$
where the unknown is the function $y=y(x)$.
````

Particular cases of the previous definition are:

1. When $f=f(x)$. In this case, the problem can be solved by integrating with respect to the independent variable $x$: 

$$
y(x)=\int f(x)\,dx\,.
$$

2. When $f=f(y)$. In this case, the ODE is called **autonomous**. It is a very common case in applications. We will later see how to solve this type of equation and some more general ones. 

````{prf:definition} Solution of a first-order ODE
:label: def_solucion_EDO 
:nonumber: 

A solution of the first-order ODE

$$
y'=f(x,y)\,,\quad x\in (a,b)\,,
$$
is any function $y=y(x)$ such that $y'(x)=f(x,y(x))$, for $x\in(a,b)$.
````

Let us look at a couple of simple examples: 

````{prf:example} 
:label: ex_EDO 
:nonumber: 

1. For the following ODE,

    $$
    y'=5y\, ,
    $$
    the function $y(x)=e^{5x}$ is a solution, since

    $$
    y'(x)=5e^{5x}=5y(x)\,.
    $$
    But... it is not the only solution! Notice that $y_C(x)=e^{5x+C}$, for any $C\in\mathbb R$, is also a solution of the previous ODE, since

    $$
    y'(x)=5e^{5x+C}=5y(x)\,.
    $$
    That is, the given ODE actually has infinitely many solutions: for each value of the constant $C$, we obtain a different solution of the ODE.
2. The solutions of the ODE,

    $$
    y'=5x\, ,
    $$
    are $\displaystyle y(x) = \frac{5 x^2}{2} + C$.

````

As we have just seen through these examples, when solving a first-order ODE the solution depends on an arbitrary constant that we may denote by $C$. To determine this constant, it is necessary to impose some additional condition. In physics and other sciences, this condition usually describes the initial state of the system, hence the name that appears in the following definition:

````{prf:definition} 
:label: def_CI_y_PI 
:nonumber: 

* An **initial condition (I.C.)** is an additional condition on the dependent variable in an ODE:

    $$
    y(x_0)=y_0\,.
    $$

* An **initial value problem (I.V.P.)** is the problem of solving an ODE together with an initial condition: 

    $$
    \left\{\begin{array}{lcl}
    y'&=&f(x,y)\,,\\
    y(x_0)&=&y_0\,.
    \end{array}\right.
    $$
````

````{prf:example} 
:label: ex_PVI 
:nonumber: 

1. The unique solution of the initial value problem

    $$
    \left\{\begin{array}{lcl}
    y'&=& 5y \, ,\\
    y(0)&=& 1 \, ,
    \end{array}\right.
    $$
    is $y(x) = e^{5x}$.
2. The unique solution of the initial value problem

    $$
    \left\{\begin{array}{lcl}
    y'&=& 5x \, ,\\
    y(0)&=& 1 \, ,
    \end{array}\right.
    $$
    is $\displaystyle y(x) = \frac{5x^2}{2}+1$.
````

A little later in this same section, we will see how to solve two special types of first-order ordinary differential equations: separable equations and first-order linear equations.


## Mathematical modeling

#### Newton's law of cooling

Newton's law of cooling (https://es.wikipedia.org/wiki/Ley_del_enfriamiento_de_Newton) states that when the difference in temperature between a body and its environment is not too large, the rate of change over time of the body's temperature is proportional to the difference between the temperature of the body and that of the surrounding medium.

We can write this idea mathematically as follows:

$$ 
\frac{dT}{dt}\,=\,-k\, (T-T_a)\,,
$$
where $T(t)$ is the temperature at time $t$, $T_a$ is the ambient temperature, and $k>0$ is a proportionality constant (its value is determined from the data of the problem). This is a first-order ordinary differential equation.

In this case, to determine the temperature of the system uniquely, we need to know the temperature at some instant of time. 

Thus, this problem is usually supplemented with an initial condition: $T(0)=T_0$, where $T_0$ is the initial temperature of the body.

Let us state this with an ODE!

We call $N(t)$ the number of radioactive nuclei at time $t$. The law of radioactive decay says that *the rate at which its nuclei decay will be constant and will not change over time*. Since this rate is the variation in the number of nuclei, that is, the derivative of $N$ with respect to time, we can write the differential equation governing the process:

$$
\frac{dN}{dt} = -kN \Leftrightarrow N' = -k N.
$$

If we know the initial amount of material, $N_{0}$, then we already have an initial value problem to work with:

$$
\left\{\begin{array}{lcl}
\displaystyle\frac{dN}{dt} &=& -kN \\
N(0) &=& N_{0}
\end{array} \right.
$$

### Newton's second law

Newton's second law (also known as the *fundamental principle of dynamics*) states that the acceleration of a particle is directly proportional to the force $F$ applied to it and inversely proportional to its mass $m$, that is,

$$
F=m\,a\,.
$$
Since, by definition, 

$$
a(t)=x''(t)\,,
$$
where $t$ denotes time and $x(t)$ represents the displacement at time $t$, we can express Newton's second law as follows:

$$
F=m\,x''\,,
$$
which is nothing other than a second-order ordinary differential equation for the displacement function: $x=x(t)$.

### Falling body in a resisting medium

If we let a body fall in a medium with friction, and assume, in a simplified way, that the only forces acting on it are gravity and friction, we obtain the following system of forces, taking the downward direction as positive:

$$
F_{\text{total}} = mg - F_{\text{friction}},
$$
where $m$ is the mass of the falling object and $g$ the universal gravitational constant (approximately equal to $9.81$).

Now, on the one hand, according to Newton's second law, $F_{\text{total}} = ma$, and, on the other hand, we accept the usual hypothesis that the friction force is opposite to the motion and depends directly proportionally on the speed of the moving body, $F_{\text{friction}} = kv$.

This gives us the following equation, which we can write as a first-order ODE in terms of the speed of the moving body:

$$
ma = mg - kv \Leftrightarrow mv' = mg - kv.
$$

**Note:** We could also write this problem as a second-order ODE in terms of the distance traveled by the falling object:

$$
mx'' = mg - kx' \, .
$$


## Using `Sympy` to solve ODEs

Below we show how `Sympy` can be used to solve ODEs. 

It is really very simple. 

1. Independent variables are defined as symbols (`sp.Symbol`), while dependent variables are defined as functions (`sp.Function`).  
2. We define the ODE with the `sp.Eq` command, making explicit the dependence of the dependent variable on the independent one. In the following example, you can see how on line 7 we write `v(x)` every time the dependent variable $v$ appears.
3. Derivatives are written inside `sp.Eq` by indicating the dependent variable and the independent variable with respect to which we differentiate. In the example below, we write $v'$ as `v(x).diff(x)`.
4. Once the ODE is defined, we solve it with the command `sp.dsolve`.
5. We can use `sp.dsolve` without any additional attributes to find the general solution, or we can include an initial condition, which we must define as `ics`, as shown in the penultimate line of the following code.

As an example, we are going to compute the velocity of a body of mass $72$ kilograms, assuming that its initial velocity is zero and that its air-resistance coefficient is $k=0.2$. 

That is, in light of what we saw in the previous section, we are going to solve the initial value problem

$$
    \left\{\begin{array}{rcl}
    72 v'&=& 72*9.81 - 0.2 v\,,\\
    v(0) &=& 0\,,
    \end{array}\right.
$$


`````python
import sympy as sp

# Independent variable
x = sp.Symbol ('x')
# Dependent variable (defined as a Function)
v = sp.Function ('v')

# We write the ODE 
eq = sp.Eq (72*v(x).diff(x), 72*9.81 - 0.2*v(x))

# We compute its general solution (this step would not be necessary, but it remains as an example)
s_general = sp.dsolve (eq)   
display (s_general)

# We compute the particular solution we are asked for
s_particular = sp.dsolve (eq, ics={v(0): 0.0}) 
display (s_particular)
`````

## Separable ODEs

````{prf:definition} Differential equation in separable variables
:label: def_EDO_separable
:nonumber: 

The differential equation

$$
y'\,=\,f(x,y) \,\,\Leftrightarrow\,\, \frac{dy}{dx}\,=\,f(x,y)
$$
is called **separable**, or an equation in **separable variables**, if $f$ is of the form:

$$
f(x,y)\,=\,g(x)\,h(y)\,,
$$
for certain functions $g$ and $h$.
````

To solve an ODE with separable variables, we *separate* the variables and integrate:

$$
\displaystyle
\frac{dy}{dx}\,=\,g(x)\, h(y) \Rightarrow 
\displaystyle
\frac{1}{h(y)}\, dy\,=\, g(x)\, dx
 \Rightarrow 
\displaystyle
\int \frac{1}{h(y)} dy\,=\, \int g(x) dx
$$ 

````{prf:example} 
:label: ex_EDO_separable
:nonumber: 

The ODE $y'=5y$ is autonomous, a particular case of an ODE in separable variables. To solve it, we separate variables and integrate:

$$
\dfrac{dy}{dx}=5y \Rightarrow \dfrac{1}{y}\,dy = 5\,dx\Rightarrow \int\dfrac{1}{y}\,dy =\int 5\,dx\Rightarrow \ln|y|=5x+C\,,
$$
where in the first step we have assumed that $y\neq 0$.
Now we solve for $y$ as a function of $x$:

$$
\ln|y(x)|=5x+C \Rightarrow |y(x)|=e^{5x+C}, \text{ with } C\in\mathbb R. 
$$
Notice that 

$$
e^{5x+C}=e^{5x}e^{C}=C'e^{5x}\,,
$$
where $C'=e^C>0$. Therefore, removing the absolute value, we have:

$$
y(x)=Ce^{5x}, \text{ with } C\neq 0\,.
$$

Notice that for $C=0$ we also obtain a solution of the original ODE: the solution $y= 0$, which we had removed when dividing by $y$ in the first step. This solution is called a **singular solution**. 
The solutions of the ODE are therefore all the functions of the form:

$$
y(x)=Ce^{5x}\,,\quad C\in\mathbb R\,.
$$
The previous expression is called the **general solution** of the ODE. For each value of $C$, we obtain a **particular solution**.
````

````{prf:example} 
:label: ex_EDO_separable_2
:nonumber: 

We now want to solve the initial value problem

$$
\left\{\begin{array}{rcl}
y'&=&5y\,,\\
y(0)&=&3\,.
\end{array}\right.
$$
We have already obtained the general solution of the ODE: 

$$
y(x)=Ce^{5x}\,,\quad C\in\mathbb R\,.
$$ 
Then we simply impose the initial condition $y(0)=3$. From the expression for the general solution,

$$
y(0)=Ce^{0}=C\,.
$$
Therefore, for $y(x)=Ce^{5x}$ to be a solution of the proposed I.V.P., $C$ must necessarily be equal to $3$. The solution of the I.V.P. is $y(x)=3e^{5x}$.
````

````{prf:example} 
:label: ex_EDO_separable_3
:nonumber: 

We want to determine the temperature of a body, knowing that initially it is at $10ºC$, that after $2$ hours it is at $15ºC$, and that the outside temperature is constant and equal to $25ºC$. To do this, we solve the following I.V.P.:

$$
\left\{\begin{array}{rcl}
T'&=&-k(T-25)\,,\\[1ex]
T(0)&=&10\,.
\end{array}\right.
$$
We begin by solving the ODE, which is autonomous and therefore has separable variables. To do this, we separate the variables and integrate:

$$
\dfrac{dT}{dt}=-k(T-25)\Rightarrow \dfrac{1}{T-25}\,dT=-k\,dt\Rightarrow \int\dfrac{1}{T-25}\,dT=-k\,\int dt\,.
$$
Solving the integrals, we get:

$$
\ln|T-25|=-kt+C \Rightarrow |T-25|=e^{-kt+C}=C'e^{-kt}\,,\quad\text{ with }C'>0\,.
$$
Then,  

$$
T(t)-25 = Ce^{-kt}\,,\quad\text{ with }C\neq 0.
$$
Notice that for $C=0$ we also have a solution of the ODE: the constant solution $T(t)=25$, which we had removed in the first step when dividing by $T-25$. In any case, this is not a solution of our I.V.P.

To determine the solution of the I.V.P., we impose the initial condition $T(0)=10$:

$$
10-25 = Ce^{0}=C\Rightarrow C=-15\,.
$$
Therefore, the solution of the proposed I.V.P. is given by 

$$
T(t)=25-15e^{-kt}\,,\quad t\ge 0.
$$

Since we also know that after $2$ hours the body is at a temperature of $15ºC$, we can use this datum to determine the value of the constant $k$. Setting $t=2$, 

$$
15=25-15e^{-2k}\Rightarrow 15e^{-2k}=10 \Rightarrow e^{-2k}=\dfrac{2}{3}\Rightarrow -2k=\ln\Big(\dfrac{2}{3}\Big)\,,
$$
thus

$$
k=-\dfrac{1}{2}\ln\Big(\dfrac{2}{3}\Big)\quad\text{ and }\quad T(t)=25-15\Big(\dfrac{2}{3}\Big)^{t/2}\,,\quad t\ge 0. 
$$
````

````{prf:example} 
:label: ex_EDO_separable_4
:nonumber: 

Finally, let us look at an ODE in separable variables that is not autonomous.

Consider the first-order ODE

$$
xy'=1+y^2\,.
$$
This ODE can be written equivalently as

$$
\dfrac{dy}{dx}=\dfrac{1+y^2}{x}=\dfrac{1}{x}(1+y^2)\,.
$$
Therefore it is an ODE in separable variables ($g(x)=\frac{1}{x}$ and $h(y)=1+y^2$). To solve it, we separate the variables and integrate:

$$
\dfrac{1}{1+y^2}\,dy = \dfrac{1}{x}\,dx\Rightarrow \arctan(y) = \ln|x|+C\,.
$$
Solving for $y$, 

$$
y(x)=\tan(\ln|x|+C)\,.
$$
````

**Note:** Notice that in the previous examples it is always possible to check whether we have obtained the correct solution... and, if not, we can use `Sympy`. Let us see it, for example, for the last example:


`````python
import sympy as sp

x = sp.symbols ('x')
y = sp.Function ('y')

# We write the ODE 
eq = sp.Eq (x*y(x).diff(x), 1+ y(x)**2)

# We compute its general solution (this step would not be necessary, but this way we complete the exercise)
s_general = sp.dsolve (eq)   
display (s_general)
`````

## First-order linear ODEs

````{prf:definition} First-order linear differential equation
:label: def_EDO_lineal
:nonumber: 

A first-order linear differential equation is an ODE of the form

$$
y'\,+\,p(x)\,y\,=\,q(x)\,,
$$
where $p$ and $q$ are functions of $x$.
````

We will now see how to solve this type of ODE. 

If we were able to find a function $\mu(x)$ such that

$$
\mu(x)\,(y'(x)\,+\,p(x)\,y(x))\,=\,(\mu(x)\,y(x))',
$$ (eq-mu)
then, multiplying the whole linear ODE from the beginning by $\mu(x)$, we would get:

$$
\mu(x)\left(y'\,+\,p(x)\,y\right) = \mu(x) q(x) \Rightarrow (\mu(x)\,y(x))' = \mu(x) q(x),
$$
from which, after integrating, 

$$
\mu(x)y(x)=\int\mu(x)q(x)\,dx \Rightarrow y(x) = \frac{\int\mu(x)q(x)\,dx}{\mu(x)},
$$
and the problem would already be solved.

Thus, the problem reduces to computing a function $\mu$ satisfying {eq}`eq-mu`.

We will not give the details, but it is not hard to check that this holds if we choose

$$
\mu(x) = \exp\left({\int p(x)\, dx}\right).
$$
The function $\mu$ is called the **integrating factor**.

````{prf:example} 
:label: ex_EDO_lineal
:nonumber: 

The ODE $y'=5y$, which we already solved in the previous section, can also be regarded as a first-order linear ODE. 

In this case, $p(x)=-5$ and $q(x)=0$ (careful! to make this identification, the coefficient of $y'$ must be equal to $1$). Therefore we can solve this ODE using the method we have just described. 

Since $p(x)=-5$, the integrating factor is

$$
\mu(x)=e^{-5x}.
$$ 
Then:

$$
y'-5y = 0 \Rightarrow e^{-5x}y' - e^{-5x}y = 0 \Rightarrow \left( e^{-5x}y \right)' = 0 \\
\Rightarrow e^{-5x} y = C.
$$
Then,

$$
y(x)=C(e^{-5x})^{-1}=Ce^{5x}\,.
$$
Of course, we have obtained the same solution as before. 
````

````{prf:example} 
:label: ex_EDO_separable_2
:nonumber: 

Consider the initial value problem

$$
\left\{\begin{array}{rcl}
y'+y&=&\sin(x)\,,\\[1ex]
y(0)&=&2\,.
\end{array}\right.
$$
The ODE is first-order linear, with $p(x)=1$ and $q(x)=\sin(x)$. 

To solve the ODE, we first compute the integrating factor:

$$
\mu(x)=e^{\int 1\,dx}=e^{x}\,.
$$
Then, multiplying the entire ODE by $\mu(x)$,

$$
y'+y=\sin(x) \Rightarrow e^{x}y'+e^{x}y = e^{x}\sin(x) \Rightarrow \left( e^{x}y \right)' = e^{x}\sin(x)
\Rightarrow e^{x} y = \int e^{x}\sin(x)dx.
$$

We compute this integral using integration by parts to obtain

$$
e^{x} y= \frac{e^x}{2}\left(\sin(x)-\cos(x)\right)+C \Rightarrow y = \dfrac{\sin(x)-\cos(x)}{2}+Ce^{-x} .
$$

We have just obtained the general solution of the first-order linear ODE. It remains to determine the particular solution of the I.V.P., that is, the value of the constant $C$ such that $y(0)=2$. Imposing this initial condition, we have:

$$
2=\dfrac{\sin(0)-\cos(0)}{2}+Ce^0=-\dfrac{1}{2}+C\Rightarrow C=\dfrac{5}{2}\,.
$$
Therefore, the solution of the proposed I.V.P. is

$$
y(x)=\frac{1}{2}\left( \sin(x)-\cos(x)+5e^{-x}\right).
$$
````

Let us now see how to solve this last problem by taking advantage of the power of `Sympy`:


`````python
import sympy as sp

x = sp.symbols ('x')
y = sp.Function ('y')

# We write the ODE 
eq = sp.Eq (y(x).diff(x) + y(x), sp.sin(x))

# We solve for its general solution (this step is not really necessary, but this way we complete the exercise)
s_general = sp.dsolve (eq)   
display (s_general)

# We compute the particular solution we are asked for
s_particular = sp.dsolve (eq, ics={y(0): 2.0}) 
display (s_particular)
`````
