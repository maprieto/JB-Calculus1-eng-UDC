3. The value, in euros, of a car $x$ years after it was bought new is given by the function

    $$
    v(x)=2000+\frac{60000}{4^{0.25 x}}.
    $$

    The *residual value* of the car is its limiting value when the number of years increases indefinitely.

      a. What was the purchase price of the new vehicle?

      b. What will its residual value be?

      c. After how many years will its value differ from the residual value by less than 1000 euros?
