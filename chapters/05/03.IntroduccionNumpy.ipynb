## Object-oriented programming

Like all Python modules, the **NumPy** library is implemented following an object-oriented programming strategy. Therefore, any data structure in Python, even the simplest one, should be understood as an object belonging to a class, and we can perform on it all the operations implemented for that class of objects. Even a number is an object of a class:


```python
a = 3.3
print(a)
print(type(a))
isinstance(a,float)
```

For example, to check the attributes and methods that we can apply to a particular object, we can use the `dir()` function, which returns a list with their names. As can be seen in the list, we can distinguish two types of attributes and methods: those with the prefix and suffix `__*__`, and those without. The first type are called **private** and usually refer to the most basic calculations that can be performed within the class to which the object belongs.


```python
dir(a)
```

For example, we can check whether one real number is greater than another in two different ways: by using the logical operator `>` or by using its private method `__ge__`:


```python
a=6.4; b=5.3
print(a.__gt__(b))
print(a > b)
```

In any case, in Python both private and public methods are used in the same way. For example:


```python
a=3.4
print(a.is_integer())
print(a.__int__())
print(int(a))
```

## Arrays of numbers

In Python there are many ways to store numerical data (or not), such as the list or tuple structures. In particular, lists can contain a finite ordered sequence of numbers (and an index can be used to access each of the elements of the list). Moreover, they are flexible enough to contain data of different kinds (a combination of integers, real numbers, lists of lists, *etc.*).

But this flexibility of lists in Python makes their computational performance quite limited. In most scientific applications in mathematics (together with their applications to different parts of computer engineering, artificial intelligence, data science, ...), real problems need operations on huge data sets and, therefore, computational speed is very important. To work efficiently on these problems, **NumPy** provides specialized functions and suitable data structures. In particular, for collections of numbers of the same type (losing some of the flexibility of lists, but gaining computational efficiency).

**One-dimensional arrays**: A one-dimensional array is an ordered collection of numbers that can be accessed by index (thus preserving order). By default, **NumPy** arrays are row arrays.

**Creating arrays and indexing**: To create a **NumPy** array of length 10 initialized with zeros, we would use the `np.zeros()` function:


```python
u = np.zeros(10)
print(u)
print(type(u))
```

The default type of the numbers contained in **NumPy** arrays is `float64` (the type stored in `np.float`). If we want to use other types, we must use the optional argument `dtype`. The type of the numbers contained in an array can be checked by accessing the `dtype` attribute of **NumPy** arrays:


```python
print(u.dtype)
w = np.zeros(5, dtype=int)
print(w)
print(type(w))
print(w.dtype)
```

What is not possible, for example, is to add a string value (that is, a value of type `string`) to a `numpy.ndarray` object, because all the elements of the array must be of the same type (or at least of a type that admits conversion) and must also have the same size. To check the size of an array, one may use the `len` function:


```python
print(len(u))
v = np.zeros(10, dtype=int)
print(u + v) # A type conversion from int64 to float64 is performed implicitly
print(u + w) # ERROR: The vectors do not have the same size!
```

A more specific way to check the dimension of an array is to use `u.shape`, which returns a tuple with the dimensions of the array:


```python
print(u.shape)
```

`shape` informs us about the size of the array in *each* direction. In the case of arrays viewed as vectors, there is only one direction, whereas in data sets with multiple indices (matrices, or $n$-dimensional tensors), `shape` would tell us the size of these data structures in each direction. For example, to create a zero integer matrix of size $2\times 3$:


```python
A =  np.zeros((2,3), dtype=int)
print(A)
print(A.shape)
```

We can change the entries of an array by indexing,


```python
print(u)
u[0] = 10.0
u[3] = -4.3
u[9] = 1.0
print(u)
```

> **IMPORTANT NOTE**: Remember that in Python index values start at 0.

Of course, there are other ways to create arrays, such as using the `ones` function to create an array containing only *ones*:


```python
w = np.ones(5)
print(w)
print(w.dtype)
```

or an array of random values:


```python
w = np.random.rand(6)
print(w)
```

or also a `numpy.array` of numbers created from a Python list of numbers:


```python
u = [4.0, 8.0, 9.0, 11.0, -2.0]
v = np.array(u)
print(v)
```

There are two other methods for creating arrays of numbers that will be useful throughout the course (and in particular when we have to plot functions with **Matplotlib**):
- `numpy.arange` 
- `numpy.linspace`

The `arange` function creates an array with consecutive integer values (in a way similar to Python's `range` function). For example, to create the row vector $\vec{u}=(0, 1, 2, 3, 4, 5)$ using `arange`,


```python
u = np.arange(6)
print(u)
print(type(u))
print(u.dtype)
```

We can check that the number $6$ is not included because the range of values starts at $0$ and the array has only six elements. To change the numerical value at which the array starts:


```python
u = np.arange(2, 6)
print(u)
```

The `linspace` function creates an array of equally spaced numbers with an initial and final value (both included) and with a prescribed size:


```python
w = np.linspace(0., 100., 6)
print(w)
print(w.dtype)
```

This `linspace` function, together with the `meshgrid` function, will be used regularly throughout the course to plot functions with **Matplotlib**.


## Functions and arithmetic with arrays

**NumPy** arrays support the basic arithmetic operations (product, sums, differences, *etc.*):


```python
a = np.array([1.0, 0.2, 1.2])
b = np.array([2.0, 0.1, 2.1])
print(a)
print(b)

# Sum of a and b
c = a + b
print(c)
```

and the product of all its elements by a scalar value,


```python
c = 10.0*a
print(c)
```

and raising all of its components to a power:


```python
a = np.array([2, 3, 4])
print(a**2)
```

The usual mathematical functions can also be applied to each of their components:


```python
# Create the vector [0, π/2, π, 3π/2]
a = np.array([0.0, np.pi/2, np.pi, 3*np.pi/2])
print(a)

# Compute the sine of each component of the vector
b = np.sin(a)
print(b)
```

The previous code computes the sine of each coefficient of the vector `a`. We should emphasize that the function being used is `np.sin`, which depends directly on the **NumPy** module. Using any other implementation of the function in other modules (for example, in the **Math** module) could lead to an error.

Of course, we could also compute the sine of each coefficient of the vector by accessing each element via its index and carrying out the calculations inside a `for` loop:


```python
b = np.zeros(len(a))
for i in range(len(a)):
    b[i] = np.sin(a[i])

print(b)
```

In this case the program is longer and harder to read. Moreover, in many cases it will be slower. The manipulation of arrays and any calculations performed on them without accessing their indices are usually known as **vectorization**. Whenever possible, using vectorization will increase the efficiency and speed of computational codes. In the final exercise of these notes, the performance of this type of technique will be analyzed.


## Slicing arrays

When working with arrays of numbers, it is common to need to extract a subset of them to create a new array. For example, taking the first three coefficients of a vector or, in the case of matrices, restricting calculations to the second column. This type of operation is called array slicing.

Let us explore this through several examples, working with an array of random values:


```python
a = np.random.rand(5)
print(a)

# Using ':' means the entire index range, that is, from 0 to (length-1)
b = a[:]
print("Slicing using '[:]' {}".format(b))

# Using '1:3' means indices 1 -> 3 (excluding 3)
b = a[1:3]
print("Slicing using '[1:3]': {}".format(b))

# Using '2:-1' means indices 2 -> the second from the end (excluding it)
b = a[2:-1]
print("Slicing using '[2:-1]': {}".format(b))

# Using '2:-2' means indices 2 -> the third from the end (excluding it)
b = a[2:-2]
print("Slicing using '[2:-2]': {}".format(b))
```

> **NOTE**: the use of index `-1` corresponds to the last element of the array. Likewise, index `-2` corresponds to the second-to-last element, *etc.*. This convention of referencing indices from the end of an array is very useful, since with these negative indices one can refer to the last coefficients of an array without having to explicitly refer to its size (or even to know it explicitly).


If what we want is to slice an array from the beginning or from the end, we must use the index syntax with '`:`'


```python
# Using ':3' means using indices from the beginning up to 3 (excluding index 3)
b = a[:3]
print("Slicing using '[:3]': {}".format(b))

# Using '4:' means indices from 4 to the end
b = a[4:]
print("Slicing using '[4:]': {}".format(b))

# Using ':' means all indices from the beginning to the end
b = a[:]
print("Slicing using '[:]': {}".format(b))
```

Slicing can also be applied to matrices:


```python
B = np.array([[1.3, 0], [0, 2.0]])
print(B)

# Extract the second row
row = B[1, :]
print(row)

# Extract the first column (stored as a row vector)
col = B[:, 0] 
print(col)
```

There are many other strategies and syntax forms related to array slicing that are beyond the scope of this brief introduction to **NumPy**. For more detailed information, see: https://docs.scipy.org/doc/numpy/reference/arrays.indexing.html


## Exercise: acceleration with NumPy in the computation of vector norms

The Euclidean norm (or *magnitude*) of a vector $\vec{x}=(x_{0},\ldots,x_{n-1})\in\mathbb{R}^{n}$ is given by

$$
\| x \| = \sqrt{\sum_{i=0}^{n-1} x_{i} x_{i}},
$$

where $x_{i}$ is the $i$-th coefficient of $\vec{x}$. In short, its norm is nothing but the calculation of the inner product of the vector with itself, followed by taking a square root. To compute the norm value one can follow different strategies: the first is to use a loop to go through all the coefficients of the vector and add the square of each coefficient, then take the sum of all these quantities and compute its square root.

To assess the computational performance of this implementation, we will take a moderately large vector (10 million entries) and compute the running time:


```python
# Create a NumPy array with 10 million random values
x = np.random.rand(10000000)
print(type(x))
```

```python
def compute_norm(x):
    norm = 0.0
    for xi in x:
        norm += xi*xi
    return np.sqrt(norm)

%time norm = compute_norm(x)
print(norm)
```

The computation time that interests us is the one appearing under the label '`Wall time`'.
